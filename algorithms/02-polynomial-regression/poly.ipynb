# Importing Libs

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyexpat.model import XML_CQUANT_PLUS
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

# Creating Random Data

```python
np.random.seed(42)
X = 2 - 3 * np.random.normal(0, 1, 100)
Y = X - 2 * (X ** 2) + 0.5 * (X ** 3) + np.random.normal(-3, 3, 100)
X = X[:, np.newaxis]
```

# Visualize Random Data

```python
plt.figure(figsize=(8,5))
plt.scatter(X, Y, alpha=0.5)
plt.grid(True,alpha=0.1)
plt.show()
```

Notice that the Data is **not** linear !

# Training the model

```python
poly_features = PolynomialFeatures(degree=3)
X_poly = poly_features.fit_transform(X)

model_poly = LinearRegression()
model_poly.fit(X_poly, Y)
```

making predictions

```python
predictions = model_poly.predict(X_poly)
```

```python
plt.scatter(X, Y, alpha=0.5)
indices = np.argsort(X.flatten())
plt.plot(X[indices], predictions[indices], alpha = 0.5,label='Polynomial (deg 3)')
plt.grid(True,alpha=0.1)
plt.legend()
plt.show()
```
