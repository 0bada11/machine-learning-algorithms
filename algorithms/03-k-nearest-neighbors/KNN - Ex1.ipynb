# I. Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import confusion_matrix, f1_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

# II. Load The Data

```python
df = pd.read_csv(r"C:\Users\LOQ\Desktop\IT\AI course\Data\diabetes.csv")
df.head()
```

# IV Data Cleaning

we should remove the zero values because its wrong value and its will effect on the result we have

```python
zero_not_accepted = ['Pregnancies','Glucose','BloodPressure','SkinThickness','Insulin','BMI']

for col in zero_not_accepted:
    df[col] = df[col].replace(0,np.nan)
    mean = int(df[col].mean(skipna=True))
    df[col] = df[col].replace(np.nan,mean)
```

```python
df.describe()
```

# V. Split The Data

```python
# split dataset
X = df[["Pregnancies","Glucose","BloodPressure","SkinThickness",
        "Insulin","BMI","DiabetesPedigreeFunction","Age"]]
y = df["Outcome"]
```

```python
X_train , X_test , y_train, y_test = train_test_split(X, y, test_size = 0.2,)
```

# VI. Scale The Data

after preparing the data : we need to make  **Feature scaling**

```python
# we only need to scale the X's
sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
```

# VII. Train KNN model

```python
# define the model
classifier = KNeighborsClassifier(n_neighbors=11, p = 2, metric='euclidean')
classifier.fit(X_train,y_train)
y_pred = classifier.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
print(f1_score(y_test, y_pred))
```
