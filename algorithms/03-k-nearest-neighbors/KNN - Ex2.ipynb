# I. Import The Libraries

```python
import numpy as np
import pandas as pd
from pyexpat import features
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

# II. Load The Data

```python
# Load the Data
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
```

# III. Explore The Data

```python
# Check basic info
df.info()

# Statistical summary
df.describe()
```

```python
df['species'] = df['target'].map({0:'setosa', 1:'versicolor', 2:'virginica'})

print(df.head())
```

## IV. Split the data

```python
features = df[["sepal length (cm)", "sepal width (cm)", "petal length (cm)", "petal width (cm)"]]
target = df['target']

X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.2,
    random_state=42,
    stratify=target
)
```

## V. Scaling Before Training

**Note** : in Scaling we scale only X "Features"

```python
scaleModel = StandardScaler()
X_train = scaleModel.fit_transform(X_train)
X_test = scaleModel.transform(X_test)
```

## VI. Training The Model

Creating the KNN model

```python
# Create KNN model
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
# Find Predictions
y_pred = model.predict(X_test)
```

## VII. Evaluation


```python
print('Accuracy:', accuracy_score(y_test, y_pred))

print('\nConfusion Matrix:\n', confusion_matrix(y_test, y_pred))
print('\nClassification Report:\n', classification_report(y_test, y_pred))
```
