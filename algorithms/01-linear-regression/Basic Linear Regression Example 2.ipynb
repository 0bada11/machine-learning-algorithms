## Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyexpat import features
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
```

## Creating Data (using numpy)

```python
np.random.seed(42)
X = np.linspace(0, 10, 100)
# we used reshape for make the data only 1 column
Y = 3 * X + np.random.randn(100) * 2
# print(X)
# print(Y)

df = pd.DataFrame(
    {
        "X": X,
        "Y": Y,
    }
)
print(df)
```

## Visualizing the Data :

```python
plt.figure(figsize=(5,5))
plt.scatter(df["X"],df["Y"],alpha=0.5)
plt.grid(True,alpha=0.1)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

we clearly see the **linear** relationship between X and Y so we can apply "**Linear regression**"

## Split the data to Training & Splitting data + Train The Model

**NOTE** : model.fit (n_samples, n_features) => n_samples **"X"** always must be **2D**

```python
# Splitting
X_train, X_test, Y_train, Y_test  = train_test_split(df[["X"]],df["Y"], test_size=0.2)

# Training
model = LinearRegression()
model.fit(X_train, Y_train)
prediction = model.predict(X_test)
plt.figure(figsize=(5,5))
plt.scatter(X_test, Y_test, c="pink",alpha=0.5)
plt.scatter(X_train, Y_train,alpha=0.5)
# when drawing the line (X_test , model.predict(X_test))
plt.plot(X_test,prediction,c="red",alpha=0.5)
plt.grid(True,alpha=0.1)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
print(model.score(X_test, Y_test))
```
