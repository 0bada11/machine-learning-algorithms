## 🧠 What is Linear Regression?
Linear Regression is a **supervised learning algorithm** used for predicting a continuous outcome variable `y` based on one input variable `x`.

It assumes a linear relationship:
\[ y = m \times x + b \]

where:
- **m** is the slope (weight or coefficient)
- **b** is the intercept (bias)

## 📊 Step 1: Create a Synthetic Dataset

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate data
np.random.seed(42)
X = 2.5 * np.random.randn(100) + 25  # study hours
y = 50 + 2 * X + np.random.randn(100) * 10  # exam score with noise

plt.scatter(X, y, color='cyan', label='Data points', alpha=0.5)
plt.title("Synthetic Data: Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True,alpha=0.2)
plt.legend()
plt.show()
```

## ⚙️ Step 2: Implement Linear Regression from Scratch (NumPy)

```python
# Compute slope (m) and intercept (b)
X_mean = np.mean(X)
y_mean = np.mean(y)

# m = cov(x,y)/var(x)
# b = y - mx

m = np.sum((X - X_mean) * (y - y_mean)) / np.sum((X - X_mean)**2)
b = y_mean - m * X_mean

print(f"Slope (m): {m:.2f}")
print(f"Intercept (b): {b:.2f}")
```

### Visualize Regression Line

```python
y_pred = m * X + b

plt.scatter(X, y, color='cyan', label='Actual data', alpha=0.5)
plt.plot(X, y_pred, color='red', label='Regression Line')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Linear Regression Fit (From Scratch)")
plt.grid(True,alpha=0.2)
plt.legend()
plt.show()
```

## 🤖 Step 3: Linear Regression using scikit-learn

```python
from sklearn.linear_model import LinearRegression

# Reshape X for sklearn
X_reshaped = X.reshape(-1, 1)

model = LinearRegression()
model.fit(X_reshaped, y)

print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)

# Predictions
y_pred_sklearn = model.predict(X_reshaped)

plt.scatter(X, y, color='cyan', label='Actual', alpha=0.5)
plt.plot(X, y_pred_sklearn, color='red', label='Regression Line')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Linear Regression (scikit-learn)")
plt.grid(True,alpha=0.2)
plt.legend()
plt.show()
```

## 🔍 Step 4: Compare Manual vs sklearn Results

```python
print(f"Manual slope: {m:.2f}, sklearn slope: {model.coef_[0]:.2f}")
print(f"Manual intercept: {b:.2f}, sklearn intercept: {model.intercept_:.2f}")
```
