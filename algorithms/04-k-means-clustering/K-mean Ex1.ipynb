# Notes:
* k-mean is unsupervised algorithm
* we dont split the data (train_test_split)
* in unsupervised learning **y** is unknown
* in unsupervised learning we **train only X**

# importing Libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import accuracy_score
```

# Generate Data

```python
X, y = make_blobs(n_samples=500, centers=3, random_state=42, n_features=2)
```

# Visualize The Data (see the blobs)

```python
plt.style.use('default')
plt.figure(figsize=(8,5))
plt.scatter(X[:, 0], X[:, 1], c=y, alpha=0.5)
plt.grid(True, alpha=0.1)
plt.show()
```

# Train the K-mean model


```python
# try K = 1,2,3 and see the difference
model = KMeans(n_clusters=3,random_state=42)
# because K-mean unsupervised algorithm we train only the Features
model.fit(X)
# find y_pred
y_pred = model.predict(X)
# find the centers
center = model.cluster_centers_
```

important note :
True labels (y):      [0, 0, 1, 1, 2, 2]
Predicted labels:    [2, 2, 0, 0, 1, 1]

Clusters are correct, but labels are permuted.

model.cluster_centers_ => array of (x,y) for each cluster

```python
df = pd.DataFrame(X, columns=['X1', 'X2'])
df["y"] = y
df["y_pred"] = y_pred
df
```

```python
plt.style.use('default')
plt.figure(figsize=(8,5))
plt.scatter(df["X1"],df["X2"], c=df["y_pred"], alpha=0.5)
plt.scatter(center[:, 0], center[:, 1], c='red' , s=70)
plt.grid(True, alpha=0.1)
plt.show()
```

```python
inertia = []
for k in range(1, 10):
    kmeans = KMeans(n_clusters=k, random_state=0)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 10), inertia, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.show()
```
