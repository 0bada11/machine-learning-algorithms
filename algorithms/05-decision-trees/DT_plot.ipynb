```python
# Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier

# Step 1: Generate 2D synthetic dataset
X, y = make_classification(
    n_samples=200,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_clusters_per_class=1,
    random_state=42
)

# Step 2: Train Decision Tree
model = DecisionTreeClassifier(max_depth=3, random_state=0)
model.fit(X, y)

# Step 3: Predict class for each sample
y_pred = model.predict(X)

# Step 4: Plot the actual vs predicted points
plt.style.use('default')
plt.figure(figsize=(8,6))
plt.scatter(
    X[:, 0], X[:, 1],
    c=y_pred, cmap='RdYlBu', edgecolor='k', s=70, label='Predicted class', alpha=0.5
)
plt.title("Decision Tree Classification Result")
plt.grid(True, alpha=0.2)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier

# Dataset
X, y = make_classification(
    n_samples=200, n_features=2,
    n_redundant=0, n_informative=2,
    n_clusters_per_class=1, random_state=42
)

# Model
model = DecisionTreeClassifier(max_depth=3, random_state=0)
model.fit(X, y)

# Grid for true decision boundaries
x_min, x_max = X[:,0].min() - 1, X[:,0].max() + 1
y_min, y_max = X[:,1].min() - 1, X[:,1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))
Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

# Plot
plt.figure(figsize=(8,6))
plt.contourf(xx, yy, Z, cmap='RdYlBu', alpha=0.4)
plt.scatter(X[:,0], X[:,1], c=y, edgecolor='k', cmap='RdYlBu', s=70)
plt.title("Decision Tree Decision Boundaries (Exact)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
