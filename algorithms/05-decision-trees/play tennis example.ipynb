# Step 0: Import libraries

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

# Step 1: Create dataset

```python
data = {
    "Outlook": ["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast",
                "Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],
    "Temp": ["Hot","Hot","Hot","Mild","Cool","Cool","Cool",
             "Mild","Cool","Mild","Mild","Mild","Hot","Mild"],
    "Humidity": ["High","High","High","High","Normal","Normal","Normal",
                 "High","Normal","Normal","Normal","High","Normal","High"],
    "Wind": ["Weak","Strong","Weak","Weak","Weak","Strong","Weak",
             "Weak","Weak","Strong","Strong","Strong","Weak","Strong"],
    "PlayTennis": ["No","No","Yes","Yes","Yes","No","Yes",
                   "No","Yes","Yes","Yes","Yes","Yes","No"]
}
df = pd.DataFrame(data)
```

- see the Dataset

```python
df
```

# Step 2: Encode categorical features
- Manual mapping to match my explanation
- or using LabelEcoder function

```python
df["Outlook"] = df["Outlook"].map({"Overcast":0,"Rain":1,"Sunny":2})
df["Temp"] = df["Temp"].map({"Cool":0,"Hot":1,"Mild":2})
df["Humidity"] = df["Humidity"].map({"High":0,"Normal":1})
df["Wind"] = df["Wind"].map({"Strong":0,"Weak":1})
df["PlayTennis"] = df["PlayTennis"].map({"No":0,"Yes":1})
```

- see the dataset after encoding}

```python
df
```

sklearn need purely numeric dataset for decision tree

# Step 3: Split features and target

```python
X = df[["Outlook","Temp","Humidity","Wind"]]
y = df["PlayTennis"]
```

# Step 4: Fit Decision Tree with entropy


```python
clf = DecisionTreeClassifier(criterion="entropy", random_state=42)
clf.fit(X, y)
```

# Step 5: Visualize the tree

```python
plt.style.use('default')
plt.figure(figsize=(16,10))
plot_tree(clf, feature_names=X.columns, class_names=["No","Yes"], filled=True, rounded=True)
plt.show()
```
